```python
import tensorflow as tf
from tensorflow import keras

num_classes = 16  # change according to your dataset

# Input
inputs = tf.keras.Input(shape=(224, 224, 5))  # 5-channel images

# --- CNN branch ---
x = tf.keras.layers.Conv2D(32, 3, activation='relu', padding='same')(inputs)
x = tf.keras.layers.MaxPooling2D()(x)
x = tf.keras.layers.Conv2D(64, 3, activation='relu', padding='same')(x)
x = tf.keras.layers.MaxPooling2D()(x)
cnn_features = tf.keras.layers.Flatten()(x)

# --- ViT branch ---
# Example: use only the first 3 channels (RGB)
vit_input = inputs[..., :3]  

# If you have a prebuilt ViT model:
# vit_features = vit_model(vit_input)
# For demo, let's just flatten the RGB input
vit_features = tf.keras.layers.Flatten()(vit_input)

# --- Combine features ---
combined = tf.keras.layers.Concatenate()([cnn_features, vit_features])
outputs = tf.keras.layers.Dense(num_classes, activation='softmax')(combined)

# Create the multi-branch model
multi_model = tf.keras.Model(inputs=inputs, outputs=outputs)

multi_model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False),
    metrics=["accuracy"]
)

multi_model.summary()
```

```python
import tensorflow as tf
import os

# Paths
train_dir = r"C:\Users\asus\Desktop\Crops_disease_detection\train"
valid_dir = r"C:\Users\asus\Desktop\Crops_disease_detection\valid"

# Parameters
batch_size = 8
img_height, img_width = 224, 224

# Load datasets (raw, before normalization)
raw_train_dataset = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=(img_height, img_width),
    batch_size=batch_size,
    shuffle=True
)

raw_valid_dataset = tf.keras.utils.image_dataset_from_directory(
    valid_dir,
    image_size=(img_height, img_width),
    batch_size=batch_size,
    shuffle=False
)

# Save class names BEFORE mapping
class_names = raw_train_dataset.class_names
num_classes = len(class_names)

# Normalize pixel values
normalization_layer = tf.keras.layers.Rescaling(1./255)
train_dataset = raw_train_dataset.map(lambda x, y: (normalization_layer(x), y))
valid_dataset = raw_valid_dataset.map(lambda x, y: (normalization_layer(x), y))

# Prefetch for performance
train_dataset = train_dataset.prefetch(tf.data.AUTOTUNE)
valid_dataset = valid_dataset.prefetch(tf.data.AUTOTUNE)

# Print info
print("Classes:", class_names)
print("Number of classes:", num_classes)
print("Training batches:", len(train_dataset))
print("Validation batches:", len(valid_dataset))
```

```python
def to_5_channels(image, label):
    # image: (224,224,3) → (224,224,5)
    extra = tf.zeros_like(image[..., :2])  # 2 extra zero channels
    image_5 = tf.concat([image, extra], axis=-1)
    return image_5, label

train_dataset = train_dataset.map(to_5_channels)
valid_dataset = valid_dataset.map(to_5_channels)
```

```python
# Path for test set
test_dir = r"C:\Users\asus\Desktop\Crops_disease_detection\test"

# Create raw dataset
raw_test_dataset = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    image_size=(img_height, img_width),
    batch_size=batch_size,
    shuffle=False   # keep order for evaluation
)

# Normalize + expand to 5 channels
def add_channels(image, label):
    # add 2 zero channels to convert RGB → 5 channels
    extra = tf.zeros_like(image[..., :2])  
    image_5ch = tf.concat([image, extra], axis=-1)
    return image_5ch, label

test_dataset = raw_test_dataset.map(lambda x, y: (normalization_layer(x), y))
test_dataset = test_dataset.map(add_channels)

# Prefetch
test_dataset = test_dataset.prefetch(tf.data.AUTOTUNE)

print("Test batches:", len(test_dataset))
```

```python
import numpy as np

# True labels
y_true = np.concatenate([y for x, y in valid_dataset], axis=0)

# Model predictions
y_pred_probs = multi_model.predict(valid_dataset)   # probabilities
y_pred = np.argmax(y_pred_probs, axis=1)           # predicted classes
y_true = np.array(y_true)                          # convert tensor to array if needed
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_auc_score, roc_curve, auc, mean_squared_error
import matplotlib.pyplot as plt

# Accuracy
print("Accuracy:", accuracy_score(y_true, y_pred))

# Precision, Recall, F1 (macro = across all classes)
print("Precision:", precision_score(y_true, y_pred, average='macro'))
print("Recall:", recall_score(y_true, y_pred, average='macro'))
print("F1 Score:", f1_score(y_true, y_pred, average='macro'))

# RMSE
rmse = mean_squared_error(y_true, y_pred, squared=False)
print("RMSE:", rmse)

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:\n", cm)
```

```python
# Load an image (ensure it is RGB)
img = cv2.imread("valid/Apple___Black_rot/0bc40cc3-6a85-480e-a22f-967a866a56a1___JR_FrgE.S 2784_270deg.JPG")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # convert BGR to RGB
img = img.astype(np.float32)
```

```python
R = img[:,:,0]
G = img[:,:,1]
B = img[:,:,2]

# Excess Green
ExG = 2*G - R - B

# Excess Red
ExR = 1.4*R - G
```

```python
ExG_norm = (ExG - ExG.min()) / (ExG.max() - ExG.min())
ExR_norm = (ExR - ExR.min()) / (ExR.max() - ExR.min())
```

```python
# Example: Assuming img is an RGB image as a NumPy array
# img shape: (height, width, 3), values in 0-255 or 0-1

import numpy as np

# Normalize to 0-1 if needed
img = img.astype(float) / 255.0  

# Compute Excess Green (ExG) and Excess Red (ExR)
ExG = 2*img[:,:,1] - img[:,:,0] - img[:,:,2]   # 2G - R - B
ExR = 1.4*img[:,:,0] - img[:,:,1]              # 1.4R - G

# Normalize for display (0-1)
ExG_norm = (ExG - ExG.min()) / (ExG.max() - ExG.min())
ExR_norm = (ExR - ExR.min()) / (ExR.max() - ExR.min())
```
